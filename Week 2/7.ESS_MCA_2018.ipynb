```python
import pandas as pd
import prince
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
from functions.ca_mca_plot import ca_plot, mca_plot
```

```python
# Load your data
df = pd.read_csv("data/ESS2018_mcadata.csv")
print("Original Data Preview:\n", df.head())
```

```python
# Select some variables to include:
select = ["prtvtfnl", "gndr", "domicil", "edulvla"]
```

```python
# Multiple Correspondence Analysis (MCA)
mca = prince.MCA(n_components=df.shape[1]-1)
mca = mca.fit(df[select])
```

```python
mca_plot(mca, df[select], title = "ESS data")
```

```python
print(mca.eigenvalues_summary)
```

```python
mca_plot(mca, df[select], title = "ESS data", plot="both")
```
