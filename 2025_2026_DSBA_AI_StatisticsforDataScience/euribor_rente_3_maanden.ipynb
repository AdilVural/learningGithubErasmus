```python
#%pip install lxml
import requests
import pandas as pd
from bs4 import BeautifulSoup
from io import StringIO  # Voeg deze import toe

# URL van jouw screenshot
URL = "https://www.global-rates.com/nl/rentestanden/euribor/2/euribor-rente-3-maanden/"

# 1. Pagina ophalen
response = requests.get(URL)
response.raise_for_status()  # geeft error als de request mislukt

# 2. HTML parseren
soup = BeautifulSoup(response.text, "html.parser")


# 3. Alle tabellen op de pagina vinden
tables = soup.find_all("table")

dfs = []
for i, table in enumerate(tables):
    # 4. Elke tabel omzetten naar DataFrame - gebruik StringIO
    df = pd.read_html(StringIO(str(table)), decimal=",", thousands=".")[0]
 # 5. Kolomnamen opschonen (eventuele lege kolomnamen)
    df.columns = [str(c).strip() for c in df.columns]

    # 6. Probeer een kolom 'Rente' of 'Rate' schoon te maken naar float
    for col in df.columns:
        if "rente" in col.lower() or "rate" in col.lower():
            df[col] = (
                df[col]
                .astype(str)
                .str.replace("%", "", regex=False)
                .str.replace(",", ".", regex=False)
                .str.strip()
            )
            df[col] = pd.to_numeric(df[col], errors="coerce")
    dfs.append(df)

    print(f"\n=== Tabel {i+1} ===")
    print(df.head())

# 7. Optioneel: alle tabellen apart opslaan
for i, df in enumerate(dfs, start=1):
    filename = f"euribor_3m_table_{i}.csv"
    df.to_csv(filename, index=False)
    print(f"Tabel {i} opgeslagen als {filename}")
```
